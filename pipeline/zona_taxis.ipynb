```python
import pandas as pd
```

```python
df = pd.read_csv('https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv')
df.head()
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg://root:root@localhost:5432/ny_taxi"
)
```

```python
print(pd.io.sql.get_schema(df, name='yellow_taxi_data', con=engine))
```

```python
df.head(n=0).to_sql(name='zones', con=engine, if_exists='replace')
```

```python
df.to_sql(name='zones', con=engine, if_exists='append')
```
